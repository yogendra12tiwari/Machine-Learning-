```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np 
%matplotlib inline 
```

```python
df = pd.read_csv('weight_height.csv')
```

```python
df
```

```python
df.head()
```

```python
plt.scatter(df['Weight'] , df['Height'])
plt.xlabel("weight")
plt.ylabel("Height")
```

```python
## divide our dataset into independent or independent 

X = df[["Weight"]]
y = df["Height"]
```

```python
from sklearn.model_selection import train_test_split 
X_train  , X_test , y_train , y_test = train_test_split(X, y , test_size=0.20 , random_state = 42)
```

```python
X.shape 
```

```python
X_train.shape , y_train.shape , X_test.shape , y_test.shape 
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
plt.scatter(X_train , y_train)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regressor = LinearRegression()
```

```python
regressor.fit(X_train , y_train)
```

```python
print("the slop or coefficent of weight" ,regressor.coef_)
print("Interceot",regressor.intercept_)
```

```python
plt.scatter(X_train , y_train)
plt.plot(X_train , regressor.predict(X_train))
```

```python
prediction of train data
predict height output = Intercept+coef(weight)
y_pred_test = 171.0 + 14.30(X_test)
```

```python
y_pred_test = regressor.predict(X_test)
```

```python
y_pred_test , y_test
```

```python
plt.scatter(X_test , y_test)
plt.plot(X_test , regressor.predict(X_test))
```

```python
from sklearn.metrics import mean_absolute_error , mean_squared_error 
from sklearn.metrics import r2_score 
mse = mean_squared_error(y_test , y_pred_test)
mae = mean_absolute_error(y_test , y_pred_test)
rmse = np.sqrt(mse)

print(mse)
print(mae)
print(rmse)
```

```python
score = r2_score(y_test , y_pred_test)
print(score)
```
